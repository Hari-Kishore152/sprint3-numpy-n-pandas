## Notebook 3 – NumPy Random Module 

### 1. Random Numbers

NumPy provides np.random for generating random values.

Random numbers are numbers generated by a process that makes their values unpredictable or appear random. NumPy provides the np.random module to generate random numbers for simulations, testing, data analysis, and machine learning.


#### Generate a Random Integer — np.random.randint()

The randint() function generates a random integer within a specified range.

```python
import numpy as np

num = np.random.randint(1, 10)

print(num)
```

#### Generate Random Decimal Numbers — np.random.rand()

The rand() function generates random decimal numbers between 0 (inclusive) and 1 (exclusive).

```python
import numpy as np

num = np.random.rand()

print(num)
```

```python
##To generate multiple random numbers:

arr = np.random.rand(5)

print(arr)
```

#### Generate Random Numbers in a Specific Range — np.random.uniform()

The uniform() function generates random floating-point numbers within a specified range.


```python
import numpy as np

arr = np.random.uniform(10, 20, 4)

print(arr)
```

#### Select Random Values — np.random.choice()

The choice() function selects random elements from a given array.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

result = np.random.choice(arr)

print(result)
```

| Function              | Purpose                                              | Example                        |
| --------------------- | ---------------------------------------------------- | ------------------------------ |
| `np.random.randint()` | Generates random integers                            | `np.random.randint(1, 10)`     |
| `np.random.rand()`    | Generates decimal values from 0 to less than 1       | `np.random.rand(5)`            |
| `np.random.uniform()` | Generates decimal values within a range              | `np.random.uniform(10, 20, 5)` |
| `np.random.choice()`  | Selects random elements from an array                | `np.random.choice(arr)`        |
| `np.random.randn()`   | Generates values from a standard normal distribution | `np.random.randn(5)`           |
| `np.random.seed()`    | Makes the random sequence reproducible               | `np.random.seed(42)`           |


### 2. Random Seed

A seed ensures reproducibility of random results.

A random seed is a starting value used by a random number generator to produce a reproducible sequence of random numbers. Using the same seed with the same random number generation process allows us to obtain the same sequence again.

A random seed helps you generate the same sequence of random numbers again, making your Python and machine learning experiments easier to reproduce.

```python
np.random.seed(42)
print(np.random.rand(3))   # Always same output if seed is fixed
```

```python
np.random.seed(42)
print(np.random.rand(3))   # Always same output if seed is fixed

np.random.seed(42)
print(np.random.rand(3))
```

#### Example Without a Random Seed

```python
import numpy as np

print(np.random.randint(1, 100, 5))
print(np.random.randint(1, 100, 5))
```

#### Example With a Random Seed

```python
import numpy as np

np.random.seed(42)

print(np.random.randint(1, 100, 5))

np.random.seed(42)

print(np.random.randint(1, 100, 5))

```

**Explanation:**

- np.random.seed(42) initializes the random number generator with the seed 42.
- The first call generates five integers.
- Resetting the seed to 42 restarts the same random sequence.
- Therefore, both outputs are identical.

Note: 42 is just an example. You can use other integer values as seeds.

**Why Is a Random Seed Important?**

Random seeds are useful for:

- Reproducibility: Obtain the same random sequence when rerunning code.
- Debugging: Reproduce results when investigating errors.
- Machine Learning: Help reproduce experiments involving random data splitting or model initialization.
- Testing: Ensure consistent test inputs across repeated runs.

### 3. Random Integers

Random integers are whole numbers generated randomly within a specified range. NumPy provides np.random.randint() to generate random integers.

```python
#Generate a Single Random Integer

import numpy as np

num = np.random.randint(1, 10)

print(num)
```

```python
import numpy as np

arr = np.random.randint(1, 100, size=5)

print(arr)
```

```python
#Generate a 2D Array of Random Integers

import numpy as np

arr = np.random.randint(1, 10, size=(2, 3))

print(arr)
```

### 4. Random Choice

Select random elements from a given array.

Example:

```python
arr = np.array([10, 20, 30, 40])
np.random.choice(arr, size=2)
```

### 5. Shuffle
Shuffling is the process of randomly changing the order of elements in an array. NumPy provides np.random.shuffle() to shuffle the elements of an array in place, meaning it modifies the original array.

Randomly shuffle elements in-place.

Example:

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

np.random.shuffle(arr)

print(arr)
```

### 6. Permutation

Returns a new shuffled array (not in-place).

```python
arr = np.array([1,2,3,4,5])
new_arr = np.random.permutation(arr)
print(new_arr)
```

### 7. Random Distributions

NumPy supports many probability distributions:

- Uniform: np.random.uniform(low, high, size)

- Normal (Gaussian): np.random.normal(mean, std, size)

- Binomial: np.random.binomial(n, p, size)

- Poisson: np.random.poisson(lam, size)
