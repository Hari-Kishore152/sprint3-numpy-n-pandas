### 1. Vectorization

Vectorization is the process of performing an operation on an entire array or Series of values at once, rather than processing each value individually using a Python loop.

In Pandas, vectorization allows you to perform operations on entire columns of a DataFrame efficiently

**Simple Example**

Suppose you have five numbers and want to add 10 to each number.

Without vectorization: You process each number individually using a loop.

With vectorization: You add 10 to the entire Pandas Series in one operation.

#### Example: Without Vectorization

```python
numbers = [10, 20, 30, 40, 50]

result = []

for num in numbers:
    result.append(num + 10)

print(result)
```

#### With Vectorization

Using Pandas:


```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

result = numbers + 10

print(result)
```

#### Vectorization with Arithmetic Operations

Pandas supports vectorized arithmetic operations on Series.

```python
import pandas as pd

df = pd.DataFrame({
    "Price": [100, 200, 300, 400],
    "Quantity": [2, 3, 4, 5]
})

df["Total"] = df["Price"] * df["Quantity"]

print(df)
```

#### Vectorization with Mathematical Operations

You can perform mathematical operations on entire columns.

```python
import pandas as pd

df = pd.DataFrame({
    "Marks": [40, 60, 80, 90]
})

df["Bonus_Marks"] = df["Marks"] + 5
df["Percentage"] = df["Marks"] / 100 * 100

print(df)
```

#### Vectorization with Conditions

You can also use vectorized conditions to classify values.

Example: Pass or Fail

Suppose students pass if their marks are at least 40.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Ravi", "Priya", "Kumar"],
    "Marks": [85, 35, 60, 25]
})

df["Result"] = df["Marks"] >= 40

print(df)
```

### 2. apply() vs Loops

Both apply() and loops are used to perform operations on data, but they differ in performance and readability.

oth apply() and loops are used to perform operations on data, but they differ in their implementation, performance, and readability.

apply() applies a function to elements of a Series or to rows or columns of a DataFrame.

Loops such as for and while process elements one at a time, giving you explicit control over the iteration.

#### Using a Loop

A loop allows you to process each value individually.

Example: Add 10 to every number


```python
import pandas as pd

df = pd.DataFrame({
    "Marks": [40, 50, 60, 70, 80]
})

result = []

for mark in df["Marks"]:
    result.append(mark + 10)

df["Updated_Marks"] = result

print(df)
```

#### Using apply()

The apply() method allows you to apply a function to values in a Series.

Example: Add 10 to every number

```python
import pandas as pd

df = pd.DataFrame({
    "Marks": [40, 50, 60, 70, 80]
})

df["Updated_Marks"] = df["Marks"].apply(lambda x: x + 10)

print(df)
```

#### Using a Custom Function with apply()

You can define your own function and pass it to apply().

Example: Classify students as Pass or Fail

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Ravi", "Priya", "Kumar"],
    "Marks": [85, 35, 60, 25]
})

def check_result(marks):
    if marks >= 40:
        return "Pass"
    else:
        return "Fail"

df["Result"] = df["Marks"].apply(check_result)

print(df)
```

| Feature       | `apply()`                                                       | Loops                                                         |
| ------------- | --------------------------------------------------------------- | ------------------------------------------------------------- |
| Purpose       | Applies a function to Series elements or DataFrame rows/columns | Repeatedly executes a block of code                           |
| Syntax        | Usually concise                                                 | Often requires more code                                      |
| Readability   | Convenient for applying existing functions                      | Useful for explicit step-by-step logic                        |
| Control       | Controlled by the function and `apply()` configuration          | Gives explicit control over iteration                         |
| Performance   | Depends on the function; often involves Python-level iteration  | Depends on the loop and operation                             |
| Complex logic | Can apply custom functions                                      | Can handle complex conditions and multiple steps              |
| Best use      | Applying a function across a Series or DataFrame                | When you need explicit iteration or more complex control flow |


### 3. Memory Optimization

Memory optimization is the process of reducing the amount of RAM used by a Pandas DataFrame without losing necessary information.

When working with large datasets, Pandas stores data in your computer's memory (RAM). If a dataset contains millions of rows and many columns, it may consume a significant amount of memory.

Memory optimization helps you:

- Reduce RAM consumption.
- Work efficiently with large datasets.
- Process data using fewer computer resources.
- Reduce the risk of running out of memory.

#### example

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40]
})

print(df.dtypes)
print(df.memory_usage(deep=True))  #Pandas provides the memory_usage() method to check how much memory a DataFrame uses.
```

Pandas may store the Age column using the int64 data type, which uses 8 bytes per value.

However, all the ages in this example **fit comfortably within the range of an 8-bit integer.**

We can potentially reduce memory consumption by changing the data type.

#### Optimize Integer Data Types

Pandas commonly uses int64 for integer columns.

An int64 value occupies 8 bytes per element, whereas an int32 value occupies 4 bytes and an int16 value occupies 2 bytes.

If your data fits within a smaller integer type's range, you may be able to reduce memory consumption.

Example

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40]
})

print("Before optimization:")
print(df.dtypes)

df["Age"] = df["Age"].astype("int8")

print("\nAfter optimization:")
print(df.dtypes)
```

#### Optimize Floating-Point Data Types

Floating-point numbers represent decimal values.

Pandas commonly uses float64, which occupies 8 bytes per value.

You can sometimes use float32, which occupies 4 bytes per value.

Example

```python
import pandas as pd

df = pd.DataFrame({
    "Temperature": [25.5, 30.2, 28.7, 32.1, 27.4]
})

print("Before optimization:")
print(df.dtypes)

df["Temperature"] = df["Temperature"].astype("float32")

print("\nAfter optimization:")
print(df.dtypes)
```

#### Use Efficient Data Types When Reading CSV Files

Another way to optimize memory is to specify appropriate data types while reading a dataset.

- Without Specifying Data Types

```python
import pandas as pd

df = pd.read_csv("employees.csv")

Pandas infers the data types from the file's contents.
```

```python
#By Specifying Data Types

df = pd.read_csv(
    "employees.csv",
    dtype={
        "Age": "int8",
        "Salary": "float32",
        "Department": "category"
    }
)
```

#### Compare Memory Before and After Optimization

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Score": [85.5, 90.2, 78.8, 92.1, 88.4],
    "Department": ["IT", "HR", "IT", "HR", "IT"]
})

before = df.memory_usage(deep=True).sum()

df["Age"] = df["Age"].astype("int8")
df["Score"] = df["Score"].astype("float32")
df["Department"] = df["Department"].astype("category")

after = df.memory_usage(deep=True).sum()

print("Memory before:", before, "bytes")
print("Memory after:", after, "bytes")
print("Memory saved:", before - after, "bytes")
```

### 4. Category Data Type

The category data type in Pandas is used to store data that belongs to a limited number of distinct categories.

For example:

- Gender: Male, Female
- Department: IT, HR, Finance
- City: Chennai, Bangalore, Hyderabad
- Status: Active, Inactive
- Rating: Low, Medium, High

Instead of treating these values as ordinary strings, Pandas can store them as categorical data.

The main advantage: The category data type can reduce memory usage when a column contains many repeated values but relatively few unique categories.

The category data type stores repeated text values more efficiently.

Useful when a column contains a limited set of repeated values.

#### Example: Creating a Category Column

Consider a DataFrame containing employee departments.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Ravi", "Priya", "Kumar", "John"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"]
})

print(df)
print(df.dtypes)
```

- Here, the Department column contains only three unique values:

IT
    
HR

Finance

However, Pandas currently stores the column using the object data type.

We can convert it to category.

```python
df["Department"] = df["Department"].astype("category")

print(df.dtypes)
```
