## Notebook 12 – String Operations

### 1. lower()

The lower() method converts all characters in a string to lowercase.

In Pandas, it is used with the .str accessor.

Syntax

df['column'].str.lower()

#### Example

```python
import pandas as pd
 
df = pd.DataFrame({
'Name': ['HARI', 'AnU', 'RaVi']
})
 
df['Name'].str.lower()
```

#### Update Column

```python

df['Name'] = df['Name'].str.lower()
print(df)
```

### 2. upper()

The upper() method converts all characters in a string to uppercase.

Syntax

df['column'].str.upper()

```python

df = pd.DataFrame({
'Name': ['hari', 'anu', 'ravi']
})
 
df['Name'].str.upper()
```

### 3. strip()

The strip() method removes unwanted spaces from the beginning and end of strings.

Syntax

df['column'].str.strip()

```python

df = pd.DataFrame({
'Name': [' Hari ', ' Anu ', ' Ravi ']
})
 
df['Name'].str.strip()
```

#### Remove Left Spaces Only

```python
df['Name'].str.lstrip()
```

#### Remove Right Spaces Only


```python
df['Name'].str.rstrip()
```

### 5. contains()

The contains() method is used to check whether a string contains a particular word, character, or pattern.

In Pandas, we use .str.contains() to search for a specific pattern in each string of a Series or DataFrame column.

It returns:

True if the pattern is found.
    
False if the pattern is not found.

#### Basic Example

```python
import pandas as pd

s = pd.Series(["Python", "Java", "Pandas", "JavaScript"])

print(s.str.contains("Java"))
```

#### Using contains() with a DataFrame

Suppose you have a dataset containing employee names.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari Kishore", "Ravi Kumar", "Priya Sharma", "Hari Prasad"]
})

print(df["Name"].str.contains("Hari"))
```

#### Filter Rows Using contains()

You can use contains() to select only the rows that match a pattern.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari Kishore", "Ravi Kumar", "Priya Sharma", "Hari Prasad"],
    "Age": [22, 25, 23, 24]
})

result = df[df["Name"].str.contains("Hari")]

print(result)
```

#### Case-Sensitive vs. Case-Insensitive Search

By default, contains() performs a case-sensitive search.


```python
import pandas as pd

s = pd.Series(["Python", "python", "PYTHON"])

print(s.str.contains("python"))
```

```python
##To ignore capitalization, use case=False.

print(s.str.contains("python", case=False))
```

#### Handling Missing Values with contains()

Suppose a column contains missing values (None or NaN).


```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", None, "Ravi", "Hari Kumar"]
})

print(df["Name"].str.contains("Hari", na=False))  #Here, na=False treats missing values as False.
```

#### Practical Example: Search for Email Providers

Suppose you want to find customers whose email addresses contain "gmail".

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Ravi", "Priya", "Kumar"],
    "Email": [
        "hari@gmail.com",
        "ravi@yahoo.com",
        "priya@gmail.com",
        "kumar@outlook.com"
    ]
})

result = df[df["Email"].str.contains("gmail", na=False)]

print(result)
```

### 6. replace()

The replace() method is used to replace existing values with new values in a Pandas Series or DataFrame.

**Important: Pandas has two commonly used replacement approaches:**

df.replace() — replaces complete values in a Series or DataFrame.
    
df["Column"].str.replace() — replaces text patterns within strings in a column.

#### Replace a Specific Value

Suppose you have a DataFrame containing employee names.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Kumar", "Ravi", "Kumar"],
    "Age": [22, 25, 23, 24]
})

df["Name"] = df["Name"].replace("Kumar", "Krishna")

print(df)
```

#### Replace Text Within a String Using str.replace()

Suppose you want to replace "Data" with "Big Data" within strings.

```python
import pandas as pd

df = pd.DataFrame({
    "Course": ["Data Science", "Data Analytics", "Data Engineering"]
})

df["Course"] = df["Course"].str.replace("Data", "Big Data")

print(df)
```

#### Replace Multiple Values

You can replace multiple values at once using a dictionary.


```python
import pandas as pd

df = pd.DataFrame({
    "City": ["Bangalore", "Chennai", "Hyderabad", "Bangalore"]
})

df["City"] = df["City"].replace({
    "Bangalore": "Bengaluru",
    "Chennai": "Madras",
    "Hyderabad": "Hyd"
})

print(df)
```

#### Replace Values in an Entire DataFrame

You can use replace() directly on a DataFrame.


```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Bangalore", "Ravi", "Kumar"],
    "City": ["Bangalore", "Chennai", "Bangalore"]
})

df = df.replace("Bangalore", "Bengaluru")

print(df)
```

#### Replace Missing Values

Suppose your dataset contains missing values represented by NaN.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Hari", "Ravi", "Kumar"],
    "Age": [22, np.nan, 24]
})

df["Age"] = df["Age"].replace(np.nan, 0)

print(df)
```

#### Replace Multiple Values with the Same Value

Suppose a dataset uses different labels for missing information, such as "Unknown", "N/A", and "Missing".

You can replace them all with pd.NA.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Unknown", "Ravi", "N/A", "Missing"]
})

df["Name"] = df["Name"].replace(
    ["Unknown", "N/A", "Missing"],
    pd.NA
)

print(df)
```

| Method                                          | Purpose                                    |
| ----------------------------------------------- | ------------------------------------------ |
| `df.replace("old", "new")`                      | Replace complete matching values           |
| `df["Name"].replace("old", "new")`              | Replace complete values in one column      |
| `df["Name"].str.replace("old", "new")`          | Replace text within strings                |
| `df.replace({"A": "B", "C": "D"})`              | Replace multiple values using a dictionary |
| `df.replace(np.nan, 0)`                         | Replace missing values with zero           |
| `df["Name"].str.replace("-", " ", regex=False)` | Replace literal characters within strings  |


### 7. Regular Expressions (Regex)

Regular Expressions (Regex) are patterns used to search, match, extract, or replace specific text in strings.

In Pandas, regular expressions are useful for cleaning text, extracting information, validating data formats, and finding patterns in DataFrame columns.

For example, Regex can help you:

Extract numbers from strings.
    
Find email addresses.
    
Remove special characters.
    
Identify strings containing specific patterns.
    
Replace unwanted characters.
    
Extract phone numbers from text.

Pandas string methods such as .str.contains(), .str.extract(), .str.findall(), and .str.replace() support regular expressions.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": [
        "Laptop123",
        "Mobile456",
        "Tablet789",
        "Computer"
    ]
})

print(df)
```

**Suppose you want to identify which product names contain numbers.**

```python
df["Has_Number"] = df["Product"].str.contains(r"\d",regex="True")
print(df)
```

- Explanation

r"\d" is a regular expression pattern.
    
\d matches any digit from 0 to 9.

.str.contains() checks whether the pattern appears anywhere in each string.
    
The result is a Boolean Series containing True or False.

The r before the string indicates a raw string, which is commonly used to write Regex patterns in Python.

| Symbol   | Meaning                                          | Example pattern | Example match     |
| -------- | ------------------------------------------------ | --------------- | ----------------- |
| `\d`     | Any digit from 0–9                               | `r"\d"`         | `5`               |
| `\D`     | Any non-digit character                          | `r"\D"`         | `A`               |
| `\w`     | A word character: letters, digits, or underscore | `r"\w"`         | `A`, `5`, `_`     |
| `\W`     | A non-word character                             | `r"\W"`         | `@`, `#`, `!`     |
| `\s`     | Whitespace                                       | `r"\s"`         | A space           |
| `\S`     | A non-whitespace character                       | `r"\S"`         | `A`               |
| `.`      | Any character except a newline, by default       | `r"A.B"`        | `ACB`             |
| `^`      | Start of a string                                | `r"^Hello"`     | `Hello World`     |
| `$`      | End of a string                                  | `r"World$"`     | `Hello World`     |
| `*`      | Zero or more repetitions                         | `r"ab*"`        | `a`, `ab`, `abb`  |
| `+`      | One or more repetitions                          | `r"ab+"`        | `ab`, `abb`       |
| `?`      | Zero or one occurrence                           | `r"colou?r"`    | `color`, `colour` |
| `{n}`    | Exactly `n` repetitions                          | `r"\d{4}"`      | `2026`            |
| `{n,m}`  | Between `n` and `m` repetitions                  | `r"\d{2,4}"`    | `26`, `2026`      |
| `[abc]`  | Any one character from the set                   | `r"[abc]"`      | `a`, `b`, `c`     |
| `[^abc]` | Any character except those listed                | `r"[^abc]"`     | `d`, `x`          |
| `\|`     | OR operator in Regex                             | `r"cat\|dog"`   | `cat`, `dog`      |


#### Find strings containing digits

```python
import pandas as pd

df = pd.DataFrame({
    "Code": ["ABC123", "XYZ", "PQR456", "LMN"]
})

df["Contains_Digit"] = df["Code"].str.contains(r"\d", na=False)

print(df)
```

#### Filter rows containing numbers

```python
result = df[df["Code"].str.contains(r"\d", na=False)]

print(result)
```

#### .str.extract() — Extract a Pattern

The .str.extract() method extracts the first match of a Regex pattern containing a capturing group.

A capturing group is created using parentheses ().

Example: Extract numbers from product codes

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop123", "Mobile456", "Tablet789"]
})

df["Number"] = df["Product"].str.extract(r"(\d+)")

print(df)
```

```python
#Note: Extracted values are usually strings. Convert them to numeric values if you need to perform mathematical calculations.

df["Number"] = pd.to_numeric(df["Number"])
```

#### .str.findall() — Find All Matches

The .str.findall() method returns a list containing all non-overlapping matches of a pattern in each string.

Example: Extract all numbers

```python
import pandas as pd

df = pd.DataFrame({
    "Text": [
        "I have 2 laptops  5 speakers 2 pens  and 3 phones",
        " I have 5 books",
        "No numbers here"
    ]
})

df["Numbers"] = df["Text"].str.findall(r"\d+")

print(df)
```

Notice the difference:

.str.extract() extracts the first match.
    
.str.findall() extracts all matches.

#### .str.replace() — Replace a Pattern

The .str.replace() method replaces text matching a Regex pattern.

Example: Remove all digits

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop123", "Mobile456", "Tablet789"]
})

df["Product"] = df["Product"].str.replace(r"\d+", "", regex=True)

print(df)
```

#### Remove special characters

```python
df = pd.DataFrame({
    "Name": ["Hari@123", "Ravi#456", "Priya!789"]
})

df["Name"] = df["Name"].str.replace(r"[^A-Za-z]", "", regex=True)

print(df)
```

#### Names Starting with H

| Pattern    | Meaning                                                   | Example Match             |
| ---------- | --------------------------------------------------------- | ------------------------- |
| `\d`       | Digit (0–9)                                               | `5`, `123`                |
| `\D`       | Non-digit                                                 | `A`, `#`, `abc`           |
| `[A-Z]`    | Uppercase letter                                          | `H`, `P`, `Z`             |
| `[a-z]`    | Lowercase letter                                          | `h`, `p`, `z`             |
| `[A-Za-z]` | Any uppercase or lowercase English letter                 | `H`, `Hari`, `ABC`, `xyz` |
| `\s`       | Whitespace character                                      | `" "`                     |
| `^`        | Matches the start of a string                             | `^H` → `Hari`             |
| `$`        | Matches the end of a string                               | `com$` → `hari@gmail.com` |
| `+`        | Matches one or more occurrences of the preceding pattern  | `\d+` → `12345`           |
| `*`        | Matches zero or more occurrences of the preceding pattern | `ab*` → `a`, `ab`, `abb`  |
