### 3. Explain Broadcasting with an Example

Broadcasting is a NumPy feature that allows arithmetic operations between arrays of different shapes by automatically expanding the smaller array.

```python
import numpy as np

arr = np.array([1, 2, 3])

result = arr + 10

print(result)
```

#### Explanation

- NumPy automatically treats:


10

as:


[10 10 10]

and performs:

[1 2 3]
+
[10 10 10]   #This automatic expansion is called Broadcasting.

### 4. Difference Between Series and DataFrame

In Pandas, Series and DataFrame are two fundamental data structures used to store, organize, and manipulate data.

Series: A one-dimensional labeled array.

DataFrame: A two-dimensional labeled data structure consisting of rows and columns. 

| Feature           | Series                                      | DataFrame                                         |
| ----------------- | ------------------------------------------- | ------------------------------------------------- |
| Definition        | One-dimensional labeled data structure      | Two-dimensional labeled data structure            |
| Structure         | Single sequence of values                   | Rows and columns                                  |
| Dimensions        | 1D                                          | 2D                                                |
| Data organization | Similar to a single column                  | Similar to a table                                |
| Column names      | Does not have DataFrame-style column labels | Has column labels                                 |
| Index             | Has an index                                | Has a row index and column labels                 |
| Data types        | Typically one data type for its values      | Different columns can have different data types   |
| Creation          | `pd.Series()`                               | `pd.DataFrame()`                                  |
| Accessing data    | `series[0]` or `series.loc[label]`          | `df["Name"]`, `df.loc[]`, `df.iloc[]`             |
| Example           | A student's marks                           | A table containing student names, ages, and marks |

### 6. Merge Two DataFrames

The merge() function is used to combine two DataFrames based on a common column, similar to SQL JOIN operations.

```python
import pandas as pd

df1 = pd.DataFrame({
    'EmpID': [1, 2, 3],
    'Name': ['Hari', 'Anu', 'Ravi']
})

df2 = pd.DataFrame({
    'EmpID': [1, 2, 3],
    'Salary': [50000, 60000, 55000]
})

result = pd.merge(df1, df2, on='EmpID')

print(result)
```

### 7. Perform GroupBy and Aggregation

groupby() is used to group data, and aggregation functions summarize each group.

```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'HR', 'IT', 'HR'],
    'Salary': [50000, 45000, 60000, 55000]
})

result = df.groupby('Department')['Salary'].agg(
    ['sum', 'mean', 'max', 'min']
)

print(result)
```

groupby() splits data into groups, and agg() applies statistical functions to summarize each group.

### 8. Filter Rows Based on Multiple Conditions

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Hari', 'Anu', 'Ravi'],
    'Age': [25, 30, 22],
    'City': ['Hyderabad', 'Delhi', 'Hyderabad']
})

filtered_df = df[
    (df['Age'] > 24) &
    (df['City'] == 'Hyderabad')
]

print(filtered_df)
```

### 9. Handle Missing Values

```python
Missing values can be identified and handled using methods such as:

- isnull()
- dropna()
- fillna()
- interpolate()
```

#### isnull() — Identify Missing Values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ravi", "Kumar", "Priya"],
    "Age": [22, np.nan, 24]
})

print(df.isnull())
```

#### dropna() — Remove Missing Values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ravi", "Kumar", "Priya"],
    "Age": [22, np.nan, 24]
})

print(df.dropna())
```

#### fillna() — Replace Missing Values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ravi", "Kumar", "Priya"],
    "Age": [22, np.nan, 24]
})

df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

#### interpolate() — Estimate Missing Values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Sales": [100, np.nan, 300, np.nan, 500]
})

df["Sales"] = df["Sales"].interpolate()

print(df)
```
