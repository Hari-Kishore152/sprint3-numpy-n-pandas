### 1. CSV Files

CSV (Comma-Separated Values) is a file format used to store tabular data in plain text, where values are typically separated by commas.

CSV files organize data into rows and columns and are commonly used in data analysis, data science, and machine learning.

Example CSV:


- Name,Age,City
- Hari,25,Chittoor
- Anu,28,Hyderabad
- Ravi,22,Bangalore


#### Writing a CSV File

```python
#We use the to_csv() method to save a DataFrame to a CSV file.

import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("CSV file created successfully!")
```

#### Reading a CSV File Using Pandas

```python
#We use pd.read_csv() to load a CSV file into a Pandas DataFrame.

import pandas as pd

df = pd.read_csv("students.csv")

print(df)
```

### 3. JSON Files

JSON stands for JavaScript Object Notation.

JSON is a lightweight data format used to store and exchange data between applications, websites, APIs, and servers.

#### Writing JSON Files

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_json("students.json")  #By default, Pandas uses an orientation that organizes the JSON data by columns.

print("JSON file created successfully!")
```

#### Understanding the orient Parameter

The orient parameter specifies how Pandas organizes the data when converting a DataFrame into JSON.

#### orient="records"

- This stores each row as a separate JSON object.

- df.to_json("students.json", orient="records", indent=4)

Output:

[

    {
        "Name": "Hari",
        "Age": 22,
        "Marks": 85
    },
    
    {
        "Name": "Ravi",
        "Age": 23,
        "Marks": 90
    },
    
    {
        "Name": "Kiran",
        "Age": 21,
        "Marks": 78
    }
    
    
]

- Explanation:

- Each {} represents one row.

- Column names become keys.

- Column values become values.

- indent=4 makes the JSON easier to read by adding indentation.

This format is useful when representing lists of records, such as employees, students, or products.

#### orient="columns"

This organizes the data by column.

#### Reading JSON Files

```python
import pandas as pd

df = pd.read_json("students.json")

print(df)
```

### 4. Parquet Files

Parquet is a columnar storage format designed for big data processing.

Apache Parquet is a column-oriented file format designed to store structured data efficiently.

It is commonly used in data engineering, big data processing, data analytics, and machine learning.

Parquet files usually have the .parquet extension.

#### Writing Parquet Files

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_parquet("students.parquet", index=False)

print("Parquet file created successfully!")
```

#### Reading Parquet Files

```python
import pandas as pd

df = pd.read_parquet("students.parquet")

print(df)
```

### 5.Pickle Files

Pickle is a Python serialization format used to save Python objects into a file and load them back later.

Serialization means converting an object into a format that can be stored or transferred. Deserialization means reconstructing the object from that stored representation.

Pickle files commonly use the .pkl or .pickle extension.

#### Write Pickle

```python
df.to_pickle("students.pkl")

print("Pickle file created.")
```

#### Read Pickle

```python
df = pd.read_pickle("students.pkl")

print(df)
```

### 7.Reading Multiple Files

Reading multiple files means loading data from several files into Python using Pandas.

For example, suppose you have three CSV files:

- students_2024.csv
- students_2025.csv
- students_2026.csv

Instead of reading each file separately, you can read multiple files and combine their data into a single DataFrame.

```python
import pandas as pd

df1 = pd.read_csv("students_2024.csv")
df2 = pd.read_csv("students_2025.csv")
df3 = pd.read_csv("students_2026.csv")

combined_df = pd.concat([df1, df2, df3], ignore_index=True)

print(combined_df)
```

#### Using glob

```python
import pandas as pd
import glob
 
files = glob.glob("sales_*.csv")
 
df = pd.concat(
[pd.read_csv(file) for file in files],
ignore_index=True
)
 
print(df)
```

**OR**

```python
import pandas as pd
import glob
import os

files = glob.glob("data/*.csv")

dataframes = []

for file in files:
    df = pd.read_csv(file)
    dataframes.append(df)

combined_df = pd.concat(dataframes, ignore_index=True)

print(combined_df)
```
