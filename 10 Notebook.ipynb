## Notebook 10 – Combining Datasets

### 1. concat()

```python
concat() is used to combine two or more DataFrames either:

Vertically (row-wise)
Horizontally (column-wise)

Syntax

pd.concat([df1, df2])
```

#### Vertical Concatenation (Default)

```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Hari', 'Anu']
})

df2 = pd.DataFrame({
    'Name': ['Ravi', 'Kiran']
})

pd.concat([df1, df2])
```

**Notice duplicate indexes.**

#### Reset Index

```python
pd.concat(
[df1, df2],
ignore_index=True
)  #ignore_index=True resets the row indexes.
```

#### Horizontal Concatenation

```python
pd.concat(
[df1, df2],
axis=1
)
```

### 2. merge()

In Pandas, merge() is used to combine two DataFrames based on a common column or key, similar to joining tables in SQL

```python
import pandas as pd

df1 = pd.DataFrame({
'EmpID': [1, 2, 3],
'Name': ['Hari', 'Anu', 'Ravi']
})

df2 = pd.DataFrame({
'EmpID': [1, 2, 3],
'Salary': [50000, 60000, 55000]
})

pd.merge(df1,df2, on='EmpID')
```

#### Types of Merge
#### Inner Join (Default)

```python
pd.merge(
df1,
df2,
on='EmpID',
how='inner'
)
```

#### Left Join

```python
pd.merge(
df1,
df2,
on='EmpID',
how='left'
)    #All rows from left DataFrame.
```

#### Right Join

```python
pd.merge(
df1,
df2,
on='ID',
how='right'
)  #All rows from right DataFrame.
```

#### Outer Join

```python
pd.merge(
df1,
df2,
on='ID',
how='outer'
)  #All rows from both DataFrames.
```

### 3. join()

join() combines DataFrames using their indexes.

It is a convenient method when joining by index instead of columns.


Syntax

df1.join(df2)

```python
df1 = pd.DataFrame({
    'Name': ['Hari', 'Anu']
})

df2 = pd.DataFrame({
    'Salary': [50000, 60000]
})

df1.join(df2)
```

#### Join with Different Types

df1.join(
df2,
how='left'
)

df1.join(
df2,
how='right'
)

df1.join(
df2,
how='outer'
)

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Hari", "Kumar"]
})

df2 = pd.DataFrame({
    "Name": ["Ravi", "Priya"]
})

result = pd.concat([df1, df2], ignore_index=True)

print(result)
```

Explanation: pd.concat() combines the rows of both DataFrames into a single DataFrame. ignore_index=True resets the index.

### 5. combine_first()

In Pandas, combine_first() combines two DataFrames by filling missing (NaN) values in the first DataFrame with corresponding non-missing values from the second DataFrame, matching by index and column labels.

```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame({
    "Name": ["Hari", "Kumar", "Ravi"],
    "Salary": [50000, np.nan, 60000]
})

df2 = pd.DataFrame({
    "Name": ["Hari", "Kumar", "Ravi"],
    "Salary": [55000, 40000, 65000]
})

result = df1.combine_first(df2)

print(result)
```

Explanation: The missing salary for Kumar is filled with 40000 from df2. Existing non-missing values in df1 are preserved.
