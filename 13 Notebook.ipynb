## Notebook 13 – Missing Values

### isnull()

The isnull() method is used to identify missing values (NaN, None, NaT) in a DataFrame.

It returns:

True -> Missing Value
False -> Non-Missing Value

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['Hari', 'Anu', None],
    'Age': [25, np.nan, 30]
})

df.isnull()
```

#### Count Missing Values

```python
df.isnull().sum()
```

#### Total Missing Values


```python

print(df.isnull().sum().sum())
```

#### Filter Rows with Missing Values

To select rows where the Age column contains missing values:

```python
print(df[df["Age"].isnull()])
```

### 2. notnull()

The notnull() function in Pandas is used to check whether values are present (not missing) in a Series or DataFrame.

It returns:

True — if the value is not missing.

False — if the value is missing (NaN, None, or NaT).

**notnull() is the opposite of isnull().**

```python
import pandas as pd

data = [10, None, 30, None, 50]

s = pd.Series(data)

print(s.notnull())
```

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", None, "Kumar"],
    "Age": [22, None, 25, 24],
    "Salary": [30000, 40000, None, 35000]
}

df = pd.DataFrame(data)

print(df.notnull())
```

#### Count Non-Missing Values

Use notnull().sum() to count the non-missing values in each column.

```python
print(df.notnull().sum())
```

#### Filter Rows Using notnull()

Suppose you want to select only rows where the Age column has a value.

```python
print(df[df["Age"].notnull()])
```

### 3. dropna()

The dropna() function in Pandas is used to remove rows or columns that contain missing values, such as NaN, None, or NaT.

#### Removing Rows with Missing Values

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kumar", "Priya"],
    "Age": [22, None, 25, 24],
    "Salary": [30000, 40000, None, 35000]
}

df = pd.DataFrame(data)
print(df)
print("\n",df.dropna())
```

#### dropna(axis=0) — Remove Rows

The axis=0 parameter specifies that you want to remove rows containing missing values.


```python
df.dropna(axis=0)
```

#### dropna(axis=1) — Remove Columns

The axis=1 parameter specifies that you want to remove columns containing missing values.

```python
print(df.dropna(axis=1))
```

#### dropna(how="any")

Removes a row if at least one value is missing.

```python
print(df.dropna(how="any"))
```

#### dropna(how="all")

Removes a row only when all values in that row are missing.

```python
data = {
    "Name": ["Hari", None, "Kumar"],
    "Age": [22, None, 25],
    "Salary": [30000, None, 35000]
}

df = pd.DataFrame(data)

print(df.dropna(how="all"))
```

#### dropna(subset=[]) — Check Specific Columns

Use subset when you want to remove rows based on missing values in particular columns.

Example: Remove rows only when the Age value is missing.

```python
print(df.dropna(subset=["Age"]))
```

```python
#You can also check multiple columns:

print(df.dropna(subset=["Age", "Salary"]))
```

### 4. fillna()

The fillna() function in Pandas is used to replace missing values (NaN, None, or NaT) with a specified value.

Instead of deleting rows containing missing values using dropna(), you can use fillna() to fill those missing values.

#### Fill Missing Values with a Specific Value

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kumar", "Priya"],
    "Age": [22, None, 25, None],
    "Salary": [30000, 40000, None, 35000]
}

df = pd.DataFrame(data)

print(df.fillna(0))
```

#### Fill Missing Values in a Specific Column

Suppose you want to fill missing values only in the Age column.

```python
df["Age"] = df["Age"].fillna(0)

print(df)
```

####  Fill Missing Values with Different Values for Different Columns

You can use a dictionary to specify a different replacement value for each column.

```python
df = df.fillna({
    "Age": 18,
    "Salary": 25000
})

print(df)
```

#### fillna() with Mean

The mean is the average of the available numerical values.

You can replace missing values with the mean of their column.


```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kumar", "Priya"],
    "Age": [22, 24, None, 26]
}

df = pd.DataFrame(data)

df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

#### fillna() with Mode

The mode is the most frequently occurring value.

It is useful for categorical data, such as city, department, or product category.

```python
import pandas as pd

data = {
    "City": ["Chittoor", "Bangalore", "Chittoor", None, "Chittoor", "Bangalore"]
}

df = pd.DataFrame(data)

mode_value = df["City"].mode()[0]

df["City"] = df["City"].fillna(mode_value)

print(df)
```

#### Forward Fill — ffill()

Forward fill replaces a missing value with the previous available value.

```python
import pandas as pd

data = {
    "Temperature": [30, None, None, 32, None]
}

df = pd.DataFrame(data)

print(df.ffill())
```

### 5. interpolate()
The interpolate() function in Pandas is used to estimate missing values based on the surrounding available values.

Instead of replacing every missing value with a fixed number, interpolate() calculates an estimated value using a method such as linear interpolation

```python
import pandas as pd

data = {
    "Temperature": [20, None, 30, None, 50]
}

df = pd.DataFrame(data)
print(df)
print("\n",df.interpolate(method="linear"))
```

#### How Does Linear Interpolation Work?

Estimates missing values using a straight-line relationship between known values.

Linear interpolation assumes that values change at a constant rate between two known points.

- For example, suppose the temperature is:

Day	Temperature

Monday	 20°C

Tuesday	  Missing

Wednesday	30°C

- The temperature increases by 10°C over two days.

Assuming a constant rate of change:

Monday: 20°C

Tuesday: 25°C

Wednesday: 30°C

Pandas estimates Tuesday's temperature as 25°C.

#### Time Interpolation

Estimates missing values based on the actual time intervals between observations.

This method requires a DatetimeIndex.

```python
import pandas as pd

dates = pd.to_datetime([
    "2026-01-01",
    "2026-01-02",
    "2026-01-05"
])

df = pd.DataFrame(
    {"Temperature": [20, None, 40]},
    index=dates
)

print(df.interpolate(method="time"))
```

#### Polynomial Interpolation

Polynomial interpolation estimates missing values using a polynomial curve rather than a straight line.

```python
df.interpolate(method="polynomial", order=2)
```

Here, order=2 specifies a second-degree polynomial.

Note: Polynomial interpolation requires sufficient valid data points. It can also produce unrealistic estimates if the chosen curve does not fit the data well.

### 6. Missing Value Handling Strategies

Missing value handling is an important step in data cleaning and preprocessing. Before using a dataset for data analysis or machine learning, we need to identify missing values and decide how to handle them.

There are several strategies for handling missing values in Pandas.

#### 1.Remove Missing Values

Use dropna() to remove rows containing missing values.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Ravi", "Kumar", "Priya"],
    "Age": [22, None, 25, 24],
    "Salary": [30000, 40000, None, 35000]
})

df_cleaned = df.dropna()

print(df_cleaned)
```

#### 2.Fill Missing Values with a Constant

Use fillna() to replace missing values with a fixed value.


```python
df["Age"] = df["Age"].fillna(0)

print(df)
```

#### 3.Mean ,median,or mode Imputation

The mean is the average of the available numerical values.

Example:

```python
df = pd.DataFrame({
    "Age": [20, 22, None, 26, 28]
})

df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

#### Forward Fill and Backward Fill

These methods use nearby observations to fill missing values.

Forward Fill — ffill()

Uses the previous available value.

```python
df = pd.DataFrame({
    "Temperature": [30, None, None, 35]
})

print(df.ffill())
```

```python
print(df.bfill())
```

#### Interpolation

Interpolation estimates missing numerical values using surrounding observations.


```python

df = pd.DataFrame({
    "Temperature": [20, None, 30, None, 50]
})

df["Temperature"] = df["Temperature"].interpolate()

print(df)
```

#### Common Missing Value Handling Strategies

| S.No. | Strategy               | Description                                                    | Pandas Function                |
| ----: | ---------------------- | -------------------------------------------------------------- | ------------------------------ |
|     1 | Remove missing rows    | Deletes rows containing missing values                         | `dropna()`                     |
|     2 | Remove missing columns | Deletes columns containing missing values                      | `dropna(axis=1)`               |
|     3 | Fill with a constant   | Replaces missing values with a fixed value                     | `fillna()`                     |
|     4 | Mean imputation        | Replaces missing numerical values with the mean                | `fillna(df["col"].mean())`     |
|     5 | Median imputation      | Replaces missing numerical values with the median              | `fillna(df["col"].median())`   |
|     6 | Mode imputation        | Replaces missing values with the most frequent value           | `fillna(df["col"].mode()[0])`  |
|     7 | Forward fill           | Uses the previous available value                              | `ffill()`                      |
|     8 | Backward fill          | Uses the next available value                                  | `bfill()`                      |
|     9 | Interpolation          | Estimates missing values using surrounding data                | `interpolate()`                |
|    10 | Group-based imputation | Fills missing values using statistics calculated within groups | `groupby()` with `transform()` |
|    11 | Missing indicator      | Creates a column indicating whether a value was missing        | `isnull()`                     |
