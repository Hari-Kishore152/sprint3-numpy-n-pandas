### 1.head()

The head() function displays the first few rows of a DataFrame.

By default, it returns the first 5 rows.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran", "Priya", "Arun", "Sneha"],
    "Age": [22, 23, 21, 24, 22, 23],
    "Marks": [85, 90, 78, 95, 88, 92]
}

df = pd.DataFrame(data)

print(df.head())
```

#### Display a Specific Number of Rows

You can specify how many rows you want to display by passing a number inside head().

```python
#Display the First 3 Rows
df.head(3)
```

### 2.tail()

The tail() function displays the last few rows of a DataFrame.

By default, it returns the last 5 rows.

#### Example


```python
df.tail()
```

#### Show Last N Rows

```python
df.tail(2)
```

### 3.sample()

The sample() function returns random rows from a DataFrame.

The sample() method in Pandas is used to select random rows or columns from a DataFrame or Series.

Unlike head(), which displays the first rows, or tail(), which displays the last rows, sample() selects data randomly.

Useful for quickly inspecting large datasets.

#### Example

```python
print(df.sample())
```

#### Select Multiple Random Rows

You can specify how many random rows you want using the n parameter.

```python
print(df.sample(n=3))
```

#### Select Random Rows Using frac

Instead of specifying the exact number of rows, you can specify the fraction of the dataset to select.

```python
#Example 1: Select 50% of the Rows
print(df.sample(frac=0.5))
```

### 4.info()

The info() function provides a summary of the DataFrame.

The info() method in Pandas provides a concise summary of a DataFrame, including its number of rows, column names, non-null value counts, data types, and memory usage.

It is useful when you want to understand the structure of a dataset before performing data analysis.

#### Example

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran", "Priya", "Arun"],
    "Age": [22, 23, 21, 24, 22],
    "Marks": [85, 90, 78, 95, 88]
}

df = pd.DataFrame(data)

df.info()
```

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", None, "Priya"],
    "Age": [22, 23, 21, None],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

df.info()
```

### 5.describe()

The describe() function generates descriptive statistics for numerical columns.

By default, describe() calculates statistics for numeric columns in a DataFrame.

The describe() method in Pandas is used to generate descriptive statistics that summarize the data in a DataFrame or Series.

It helps you understand your dataset by calculating statistical measures such as:

Count
Mean
Standard deviation
Minimum value
Maximum value
Quartiles (25%, 50%, and 75%)

#### Example

```python
print(df.describe())
```

#### Notice: The Name column is not included because it contains text. By default, describe() summarizes numeric columns.

#### Describe Specific Columns

Suppose you want to calculate statistics only for the Marks column.

```python
print(df["Marks"].describe())
```

#### Describe All Columns, Including Text

By default, describe() summarizes numeric columns.

If you want to include text columns as well, use:

```python
print(df.describe(include="all"))
```

### 9.memory_usage()

The memory_usage() function shows how much memory each column consumes.

It helps identify columns that take large amounts of memory.

By default, it returns the memory usage of each column in bytes, including the DataFrame's index.

```python
df.memory_usage()
```

- Index represents the memory used by the DataFrame's index.
- Name represents the memory attributed to that column's underlying data.
- Age represents the memory attributed to the age values.
- Marks represents the memory attributed to the marks values.

Important: By default, Pandas does not include the full memory consumed by Python objects stored in object-type columns. We can request a deeper estimate using deep=True.

#### Understanding deep=True

When a column contains strings or other Python objects, the default memory calculation may not include the memory used by the objects themselves.

Use deep=True for a more comprehensive estimate.

Include Actual Object Memory

For string columns:


```python
print(df.memory_usage(deep=True))
```

#### total mem usage

```python
total_memory = df.memory_usage(deep=True).sum()

print("Total memory usage:", total_memory, "bytes")
```

#### Display Memory Usage in Kilobytes

Memory usage is often easier to understand in kilobytes (KB).

You can divide the total number of bytes by 1024.

```python
total_memory = df.memory_usage(deep=True).sum()

memory_kb = total_memory / 1024

print("Total memory usage:", round(memory_kb, 2), "KB")
```

### 10.nunique()

In Pandas, nunique() is used to count the number of unique (distinct) values in a Series or DataFrame column.

In simple words, it tells you how many different values exist in a column, without counting repeated values multiple times.

```python
import pandas as pd

data = {
    "Student": ["Ravi", "Priya", "Kumar", "Sneha", "Arun"],
    "Department": ["CSE", "ECE", "CSE", "IT", "ECE"]
}

df = pd.DataFrame(data)

print(df)

print("\n",df.nunique())

```

#### single column

```python

print("\n Number of unique departments:", df["Department"].nunique())
```

### 11.value_counts()

In Pandas, value_counts() counts how many times each unique value appears in a column.

The value_counts() function counts the frequency of each unique value in a column.

```python
df = pd.DataFrame({
    'Department': [
        'IT',
        'HR',
        'IT',
        'Finance',
        'IT',
        'HR'
    ]
})

print(df['Department'].value_counts())
```

#### Include Percentage Distribution

```python

df['Department'].value_counts(normalize=True)
```

#### Include Missing Values

```python
df['Department'].value_counts(dropna=False)
```
