### 1. Datetime

The Python datetime module is used to work with dates and times. It allows you to get the current date and time, extract the year or month, calculate differences between dates, and format dates into strings.

In Pandas, Datetime is a data type used to represent dates and times.

**Pandas stores datetime values using the datetime64[ns] data type.**

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Kumar", "Ravi", "Priya"],
    "Joining_Date": ["2026-01-15", "2026-02-20", "2026-03-10", "2026-04-25"]
})
print(df.dtypes)
```

**Currently, the Joining_Date column contains strings. We need to convert them into Pandas datetime format to perform date-related operations.**

#### 1. pd.to_datetime()

Converts a column or value into datetime format.

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

print(df["Joining_Date"])

print("\n",df.dtypes)
```

#### Extract Year, Month, and Day

Use the .dt accessor to extract information from a datetime column.

```python
#Extract year
df["Year"] = df["Joining_Date"].dt.year
print(df)
```

```python
#Extract month
df["Month"] = df["Joining_Date"].dt.month
print(df)
```

```python
#Extract day
df["Day"] = df["Joining_Date"].dt.day
print(df)
```

#### Extract Day of the Week

Use .dt.day_name() to get the weekday name.


```python
df["Day_Name"] = df["Joining_Date"].dt.day_name()

print(df[["Name", "Joining_Date", "Day_Name"]])
```

#### Extract Month Name

Use .dt.month_name().

```python
df["Month_Name"] = df["Joining_Date"].dt.month_name()

print(df[["Joining_Date", "Month_Name"]])
```

#### Filter Rows Using Dates

Suppose you want employees who joined after February 2026.


```python
result = df[df["Joining_Date"] > "2026-02-01"]

print(result)
```

#### You can also filter within a date range.

```python

result = df[
    (df["Joining_Date"] >= "2026-02-01") &
    (df["Joining_Date"] <= "2026-03-31")
]

print(result)
```

### 3. Date Parsing

Date parsing is the process of converting date and time values stored as strings or other formats into a proper datetime format that Pandas can understand and work with.

Date Parsing is the process of converting strings into datetime objects.

Pandas uses to_datetime() for this purpose.

**Date parsing is useful when working with datasets containing dates in different formats, such as 25-12-2026, 2026/12/25, or December 25, 2026.**

#### Parsing Different Date Formats

Dates can appear in different formats depending on the dataset.


```python
import pandas as pd

df = pd.DataFrame({
    "Date": ["25-12-2026", "15-01-2026", "10-03-2026"]
})

df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")

print(df)

# The parameter format="%d-%m-%Y" tells Pandas how to read the input dates.

# After conversion, Pandas displays the datetime values in its standard format, YYYY-MM-DD.
```

**Common format codes**

| Code | Meaning                | Example    |
| ---- | ---------------------- | ---------- |
| `%d` | Day                    | `25`       |
| `%m` | Month                  | `12`       |
| `%Y` | Four-digit year        | `2026`     |
| `%y` | Two-digit year         | `26`       |
| `%H` | Hour (24-hour format)  | `14`       |
| `%M` | Minute                 | `30`       |
| `%S` | Second                 | `45`       |
| `%B` | Full month name        | `December` |
| `%b` | Abbreviated month name | `Dec`      |


#### Handling Invalid Dates

Sometimes datasets contain invalid date values.

For example, "invalid_date" cannot be interpreted as a valid date.

Use errors="coerce" to convert invalid values into NaT (Not a Time).

```python
import pandas as pd

df = pd.DataFrame({
    "Date": ["2026-01-15", "invalid_date", "2026-03-10"]
})

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print(df)
```

#### How to actually change the format

If you want to convert 2026-01-15 into 15-01-2026, use .dt.strftime().

```python
import pandas as pd

df = pd.DataFrame({
    "Date": ["2026-01-15", "invalid_date", "2026-03-10"]
})

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Date"] = df["Date"].dt.strftime("%d-%m-%Y")

print(df)
```

### 4. Timedelta

In Pandas, Timedelta represents a duration or the difference between two dates or times. It is used to measure how much time has passed or to add or subtract time from a date

For example:

Difference between two dates: 10 days.
    
Adding 5 days to a date.
    
Subtracting 2 hours from a timestamp.

Calculating how long an employee has worked.

Pandas provides pd.Timedelta() to create a duration.

#### Creating a Timedelta

You can create a duration using pd.Timedelta().

```python
import pandas as pd

td = pd.Timedelta(days=5)

print(td)
print(type(td))
```

```python
#You can specify different time units:

import pandas as pd

print(pd.Timedelta(days=5))
print(pd.Timedelta(hours=10))
print(pd.Timedelta(minutes=30))
print(pd.Timedelta(seconds=45))
print(pd.Timedelta(weeks=2))
```

#### Adding Timedelta to a Timestamp

You can add a duration to a date or timestamp to calculate a future date.

```python
import pandas as pd

date = pd.Timestamp("2026-01-15")

new_date = date + pd.Timedelta(days=10)

print(new_date)
```

#### Finding the Difference Between Two Dates

One of the most important uses of Timedelta is calculating the difference between two dates.

```python
import pandas as pd

start_date = pd.Timestamp("2026-01-01")
end_date = pd.Timestamp("2026-01-15")

difference = end_date - start_date

print(difference)
print(type(difference))
```

#### Timedelta in a Pandas DataFrame

Suppose you have employee joining dates and want to calculate a date 30 days after each employee joins.


```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Kumar", "Ravi"],
    "Joining_Date": pd.to_datetime([
        "2026-01-01",
        "2026-02-10",
        "2026-03-15"
    ])
})

df["After_30_Days"] = df["Joining_Date"] + pd.Timedelta(days=30)
print(df)
```

#### Extract Days from a Timedelta

When you subtract two datetime columns, the result is a duration. You can use .dt.days to extract the number of whole days.


```python
import pandas as pd

df = pd.DataFrame({
    "Start_Date": pd.to_datetime(["2026-01-01", "2026-02-01"]),
    "End_Date": pd.to_datetime(["2026-01-10", "2026-02-20"])
})

df["Duration"] = df["End_Date"] - df["Start_Date"]

df["Duration_Days"] = df["Duration"].dt.days

print(df)
```

### 5. Time Zones

A time zone represents the local time used in a particular geographical region.

n Pandas, time zone functionality allows you to work with datetime values associated with specific time zones, convert times between different regions, and compare timestamps across locations.

#### Creating a Timestamp with a Time Zone

Use pd.Timestamp() with the tz parameter to create a timestamp associated with a time zone.

```python
import pandas as pd

ts = pd.Timestamp("2026-10-26 10:00:00", tz="Asia/Kolkata")

print(ts)
```

#### Getting the Current Time in a Specific Time Zone

Use pd.Timestamp.now() with the tz parameter.

```python

import pandas as pd

india_time = pd.Timestamp.now(tz="Asia/Kolkata")

print(india_time)
```

```python
###To get the current time in New York:

new_york_time = pd.Timestamp.now(tz="America/New_York")

print(new_york_time)
```

#### . Localizing a Timestamp Using tz_localize()

tz_localize() assigns a time zone to a datetime value that does not already have one.

Suppose you have a timestamp without a time zone:

```python
import pandas as pd

ts = pd.Timestamp("2026-10-26 10:00:00")

print(ts)
```

```python
#Now, assign the Indian time zone:

ts_india = ts.tz_localize("Asia/Kolkata")

print(ts_india)

# The clock time remains 10:00 AM. Pandas has associated that time with India's time zone.

# Important: tz_localize() does not convert the time from one region to another. It assigns a time zone to a timestamp that previously had none.
```

#### Converting Time Zones Using tz_convert()

tz_convert() converts a time zone-aware timestamp from one time zone to another while preserving the same actual moment in time.

For example:


```python
import pandas as pd

ts = pd.Timestamp("2026-10-26 10:00:00", tz="Asia/Kolkata")

new_york_time = ts.tz_convert("America/New_York")

print("India:", ts)
print("New York:", new_york_time)
```

### 6. Extracting Date Features

Extracting date features means extracting useful information from a date or datetime column, such as the year, month, day, weekday, quarter, hour, and minute.

In Pandas, we use the .dt accessor to extract date and time components from a datetime Series.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Hari", "Kumar", "Ravi", "Priya"],
    "Joining_Date": [
        "2026-01-15 10:30:45",
        "2026-02-20 14:15:30",
        "2026-03-10 09:45:20",
        "2026-04-25 18:20:10"
    ]
})

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

print(df)
```

#### Extracting the Year

Use .dt.year to extract the year.

```python
df["Year"] = df["Joining_Date"].dt.year

print(df[["Name", "Joining_Date", "Year"]])
```

#### Extracting the Month

Use .dt.month to extract the month as a number from 1 to 12.

```python
df["Month"] = df["Joining_Date"].dt.month

print(df[["Joining_Date", "Month"]])
```

#### Extract the Month Name

Use .dt.month_name().

```python
df["Month_Name"] = df["Joining_Date"].dt.month_name()

print(df[["Joining_Date", "Month_Name"]])
```

#### Extracting the Day

Use .dt.day to extract the day of the month.

```python
df["Day"] = df["Joining_Date"].dt.day

print(df[["Joining_Date", "Day"]])
```

#### Extracting the Day of the Week

Use .dt.day_name() to obtain the full weekday name.

```python
df["Day_Name"] = df["Joining_Date"].dt.day_name()

print(df[["Joining_Date", "Day_Name"]])
```

#### Extracting the Quarter

A year has four quarters, with three months in each quarter.

```python
df["Quarter"] = df["Joining_Date"].dt.quarter

print(df[["Joining_Date", "Quarter"]])
```

#### Extracting Hours, Minutes, and Seconds
