### 1. loc

In Pandas, loc[] is used to access and select rows and columns of a DataFrame using their labels.

In simple words, loc[] helps you retrieve specific data from a DataFrame by specifying row labels and column names

loc (label-based location) is used to select rows and columns using row labels and column names.

Syntax of loc[]

df.loc[row_label , column_label]

#### example

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Kumar", "Sneha", "Arun"],
    "Age": [22, 23, 21, 24, 22],
    "Department": ["CSE", "ECE", "IT", "CSE", "ECE"],
    "Salary": [30000, 35000, 28000, 40000, 32000]
}

df = pd.DataFrame(data)

print(df)
```

#### Selecting a Single Row Using loc[]

```python
print(df.loc[0])
```

#### Selecting a Specific Value Using loc[]

```python
print(df.loc[1,"Age"])
```

#### Selecting Multiple Rows Using loc[]

Suppose you want to retrieve the rows labeled 0, 1, and 2.

```python
print(df.loc[0:2])    #here the last value is also included
```

#### Selecting Multiple specific Rows

```python
print(df.loc[[0,2,4,]])
```

#### Selecting Specific Columns

```python
print(df.loc[:,"Name"])
```

```python
print(df.loc[:,["Name","Age"]])
```

#### Selecting Specific Rows and Columns Together

```python
print(df.loc[[0,2],["Name","Age"]])
```

#### Filtering Rows Using Conditions with loc[]

```python
print(df.loc[df["Age"] > 22])
```

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kumar", "Sneha"],
    "Age": [22, 24, 25, 23]
}

df2 = pd.DataFrame(data)

print(df2.loc[(df["Age"] > 22) & (df2["Name"] == "Kumar")])
```

#### Updating Values Using loc[]

```python
df.loc[0,"Salary"] = 52348
print(df)
```

```python
df.loc[df["Salary"] < 30000, "Salary"] = 30000

print(df)

#Pandas identifies rows where the salary is below 30000 and updates only those salary values.
```

### 2.iloc

iloc (integer location) is used to select rows and columns using integer positions (indexes).

### 3. Boolean Indexing

Boolean Indexing is a technique in Pandas used to filter rows from a DataFrame based on a condition.

It returns only the rows where the condition evaluates to True.

df["Age"] > 22

This checks the age of every employee and returns True or False.

True → The condition is satisfied.

False → The condition is not satisfied.

#### Example

```python
import pandas as pd

data = {
    "Name": ["Hari", "Kumar", "Ravi", "Priya", "Suresh"],
    "Age": [22, 25, 23, 20, 28],
    "Salary": [25000, 35000, 30000, 22000, 45000],
    "City": ["Chittoor", "Chennai", "Bangalore", "Hyderabad", "Chennai"]
}

df = pd.DataFrame(data)

print(df)

```

#### Apply a Boolean Condition

```python
print(df["Age"] > 22)
```

#### Filter the DataFrame Using Boolean Indexing

To retrieve only the employees whose age is greater than 22:

```python
df[df["Age"] > 22]
```

#### Select employees whose age is greater than 22 AND whose salary is greater than 30000.

```python
print(df[(df["Age"] > 22) & (df["Salary"] > 30000)])
```

#### Select employees whose age is greater than 25 OR whose salary is greater than 40000.

```python
print(df[(df["Age"] > 25) | (df["Salary"] > 40000)])
```

#### Select employees whose age is  not greater than 22.


```python
print(df[~(df["Age"] > 22)])
```

### 4. Filtering Rows

Filtering rows means selecting only those rows that satisfy specific conditions.

This is one of the most common operations in data analysis.

#### Filter by Single Condition

```python
df[df['Age'] > 25]
```

#### Filter by City

```python
df[df['City'] == 'Mumbai']
```

#### Using AND (&)

Filter employees whose age is greater than 22 and salary is greater than 30000.


```python
print(df[(df["Age"] > 22) & (df["Salary"] > 30000)])
```

#### Using OR (|)

Filter employees whose age is less than 22 or salary is greater than 40000.

```python
print(df[(df["Age"] < 22) | (df["Salary"] > 40000)])
```

#### Using NOT (~)

Filter employees whose age is not greater than 22.

```python
print(df[~(df["Age"] > 22)])
```

#### Filter Rows Using isin()

The isin() method filters rows when a column contains any value from a specified list.

**Select employees who live in Chennai or Hyderabad.**

```python
print(df[df["City"].isin(["Chennai", "Hyderabad"])]) 
```

```python
print(df[df["Age"].isin([22, 25, 28])])  #This selects employees whose ages are 22, 25, or 28.
```

#### Filter Rows Using between()

The between() method selects rows where values fall within a specified range. Both endpoints are included by default.

**Select employees whose age is between 22 and 25.**

```python
print(df[df["Age"].between(22, 25)])  #This selects employees aged 22, 23, 24, or 25, if those ages exist in the DataFrame.
```

#### Filter Rows Based on Missing Values

Pandas provides isna() and notna() to filter rows based on missing values.

These methods are useful when cleaning datasets.

```python
# Rows where Salary is missing
print(df[df["Salary"].isna()])
```

```python
# Rows where Salary is not missing
print(df[df["Salary"].notna()])
```

### 5.query()

In Pandas, the query() method is used to filter rows from a DataFrame by writing conditions inside a string.

It provides an alternative to Boolean indexing and can make filtering code easier to read

Syntax

df.query('condition')

```python
print(df.query("Age > 22"))
```

#### Use Multiple Conditions in query()

You can combine multiple conditions using and, or, and not.

- Example 1: Using and

Select employees whose age is greater than 22 and salary is greater than ₹30,000.

```python
print(df.query("Age > 22 and Salary > 30000"))
```

#### Filter Rows Using String Columns

You can filter rows using text values as well as numerical values.

Example 1: Select employees from Chennai

```python
print(df.query("City == 'Chennai'"))
```

### 7. between()

In Pandas, between() is used to check whether values in a Series fall within a specified range. It returns True for values within the range and False for values outside it.

The between() method checks whether values lie within a specified range.

Both boundary values are included by default.

df['column'].between(start, end)

```python
df = pd.DataFrame({
    'Age': [18, 22, 25, 30, 35]
})

df['Age'].between(20, 30)
```

#### Filter Rows using between()

```python
df[df['Age'].between(20, 30)]
```

#### 8.Sorting Data

Sorting arranges data in ascending or descending order.

In Pandas, sorting data means arranging rows in a particular order based on the values in one or more columns.

Pandas provides two important methods for sorting:

- sort_values() — Sort rows based on column values.
- sort_index() — Sort rows or columns based on their index labels.

#### sort_values()

The sort_values() method sorts a DataFrame according to the values in a specified column or columns.

Syntax

df.sort_values(by="column_name", ascending=True)

Parameters:

by — Specifies the column or columns used for sorting.
    
ascending=True — Sorts from smallest to largest or A to Z.
    
ascending=False — Sorts from largest to smallest or Z to A.

**example**

```python
print(df.sort_values(by="Age"))
```

**Sort by Age in Descending Order**

```python
print(df.sort_values(by="Age", ascending=False))
```

#### Sort Using Multiple Columns

City → Ascending.

Age → Descending.

Important: Each value in ascending corresponds to the column in the same position in by.

```python
print(df.sort_values(by=["City", "Age"], ascending=[True, False]))
```

#### sort_index()

The sort_index() method sorts a DataFrame based on its index labels rather than the values in a column.

```python
print(df.sort_index())
```

This arranges the rows according to index labels from smallest to largest.

Our original DataFrame is already in this order, so the output remains unchanged.

#### Sort by Index in Descending Order

```python
print(df.sort_index(ascending=False))
```

#### Sort Columns by Their Names

```python
print(df.sort_index(axis=1))

#Here, axis=1 sorts the column labels alphabetically.
```

```python
axis=0 → Sort rows by their index labels.
    
axis=1 → Sort columns by their labels.
```
