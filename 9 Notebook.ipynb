### 1. groupby()

The groupby() method is used to split data into groups based on one or more columns and then perform operations on each group.

It is similar to SQL's GROUP BY.

Syntax

df.groupby('column_name')

```python
import pandas as pd

data = {
    "Name": ["Hari", "Kumar", "Ravi", "Priya", "Suresh", "Divya", "Arun"],
    "Department": ["IT", "HR", "IT", "HR", "IT", "Finance", "Finance"],
    "Salary": [25000, 35000, 30000, 28000, 45000, 40000, 32000],
    "Age": [22, 25, 23, 24, 28, 26, 23]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.groupby("Department")["Salary"].sum()
```

#Maximum Salary by Department

print(df.groupby("Department")["Salary"].max())

```python
grouped = df.groupby("Department")

print(grouped)
```

#### Group by Multiple Columns

```python
df.groupby(
['Department', 'Location']
)['Salary'].sum()
```

#### Common Aggregations

df.groupby('Department')['Salary'].mean()

df.groupby('Department')['Salary'].max()

df.groupby('Department')['Salary'].min()

df.groupby('Department')['Salary'].count()

### 2. agg()

The agg() (aggregate) method is used to apply one or more aggregation functions to grouped data.

The main advantage of agg() is that you can perform multiple calculations on the same column or different columns using a single statement.

Syntax

df.groupby('column').agg(function)

#### Single Aggregation

```python
df.groupby('Department')['Salary'].agg('mean')
```

#### Multiple Aggregations

```python
df.groupby('Department')['Salary'].agg(
['sum', 'mean', 'max', 'min']
)
```

### 3. transform()

In Pandas, transform() is used to perform an operation on data and return a result that has the same number of rows as the original data.

It is especially useful when you want to calculate something for each group and add the result back to every row belonging to that group.

**Unlike agg(), it does not reduce rows.**

- Syntax

df.groupby("column_name")["numeric_column"].transform("function")

```python
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'HR'],
    'Salary': [50000, 60000, 45000, 55000]
})

df['Dept_Avg'] = df.groupby('Department')['Salary'].transform('mean')

print(df)
```

### 4. filter()

n Pandas, filter() is used with groupby() to select or keep specific groups based on a condition. It returns the original rows belonging to the groups that satisfy the condition.

The filter() method removes entire groups based on a condition

Syntax

df.groupby('column').filter(function)

**Suppose you want to select only departments whose average salary is greater than ₹50,000.**

```python
import pandas as pd

data = {
    "Name": ["Hari", "Kumar", "Ravi", "Priya", "Suresh", "Divya"],
    "Department": ["IT", "HR", "IT", "HR", "Finance", "Finance"],
    "Salary": [50000, 40000, 60000, 45000, 55000, 65000]
}

df = pd.DataFrame(data)

result = df.groupby("Department").filter(
    lambda group: group["Salary"].mean() > 50000
)

print(result)
```

```python
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'Finance'],
    'Salary': [50000, 60000, 45000, 70000]
})

df.groupby('Department').filter(
    lambda x: len(x) > 1
)
```

### 5. Pivot Tables

In Pandas, pivot_table() is used to summarize and analyze data by grouping rows and columns and applying an aggregation function such as sum(), mean(), or count().

Suppose you want to calculate the total sales for each product in each region.

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Mobile", "Laptop", "Mobile", "Laptop", "Mobile"],
    "Region": ["East", "East", "West", "West", "East", "West"],
    "Sales": [50000, 20000, 60000, 25000, 30000, 15000]
}

df = pd.DataFrame(data)

pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Product",
    columns="Region",
    aggfunc="sum",
    fill_value=0
)

print(pivot)
```

### 6. Crosstab

A Crosstab (Cross Tabulation) calculates frequency counts between two or more categorical variables.

It is commonly used for categorical data analysis.

Syntax

pd.crosstab(row, column)

```python
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'HR', 'IT'],
    'Gender': ['M', 'F', 'M', 'F', 'M']
})

pd.crosstab(
    df['Department'],
    df['Gender']
)
```
