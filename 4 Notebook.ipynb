## Notebook 4 – Introduction to Pandas


### 1. What is Pandas? 

Pandas is an open-source Python library used for data manipulation and data analysis. It helps us organize, clean, filter, modify, and analyze structured data in rows and columns.

#### Pandas provides two main data structures:

- Series: A one-dimensional labeled data structure.
- DataFrame: A two-dimensional tabular data structure consisting of rows and columns.
  
**Built on top of NumPy, optimized for handling structured/tabular data.**

**What Is Pandas Used For?**
- Reading data from CSV, Excel, and JSON files.
- Cleaning missing or incorrect data.
- Selecting and filtering rows and columns.
- Sorting data.
- Performing calculations and statistical analysis.

### 2. Why Pandas?

Pandas is used to manipulate, clean, organize, and analyze structured data efficiently in Python.

### 3. Series

A Series in Pandas is a one-dimensional labeled data structure that can store data such as integers, floating-point numbers, strings, and other Python objects.

A Series contains two main components:

- Values: The actual data stored in the Series.
- Index: Labels used to identify each value.

```python
import pandas as pd
s = pd.Series([10, 20, 30, 40], index=['a','b','c','d'])
print(s)
```

#### Accessing Series Elements

We can access values using their index labels or positions.

```python
import pandas as pd

s = pd.Series([85, 90, 78], index=["Hari", "Ravi", "Kiran"])

print(s["Hari"])
print(s.iloc[1])
```

#### Series from a Dictionary

We can also create a Series from a Python dictionary.

```python
import pandas as pd

data = {
    "Maths": 90,
    "Science": 85,
    "English": 88
}

s = pd.Series(data)

print(s)
```

### 4. DataFrame
A DataFrame is a two-dimensional, labeled data structure in Pandas that organizes data into rows and columns, similar to a table in Excel or a database.

A DataFrame consists of:

- Rows: Represent individual records.
- Columns: Represent different attributes or variables.
- Index: Identifies each row.
- Column names: Identify each column.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

#### Accessing Columns

We can select a particular column using its name.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df["Name"])
```

#### Accessing Rows

We can access rows using .loc[] and .iloc[].


**Using .loc[]**

.loc[] selects rows using their index labels.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df.loc[1])
```

**Using .iloc[]**

.iloc[] selects rows using their integer positions.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df.iloc[1])
```

#### Adding a New Column

```python
df["City"] = ["Chittoor", "Bangalore", "Hyderabad"]

print(df)
```

#### Filtering Data

```python
result = df[df["Marks"] > 80]

print(result)
```

### 5. Creating DataFrames

#### Creating a DataFrame Using a Dictionary

A dictionary contains key-value pairs. In Pandas, dictionary keys become column names, and their values become the column data.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

#### Creating a DataFrame Using a List of Lists

Each inner list represents one row in the DataFrame.

```python
import pandas as pd

data = [
    ["Hari", 22, 85],
    ["Ravi", 23, 90],
    ["Kiran", 21, 78]
]

df = pd.DataFrame(data, columns=["Name", "Age", "Marks"])

print(df)
```

#### From NumPy arrays:

```python
import numpy as np
arr = np.array([[10,20],[30,40]])
df = pd.DataFrame(arr, columns=['X','Y'])
print(df)
```

### 6. Data Types in Pandas

Data types in Pandas specify the kind of data stored in a Series or DataFrame column, such as integers, decimal numbers, text, or Boolean values.

| Data Type        | Description                                     | Example             |
| ---------------- | ----------------------------------------------- | ------------------- |
| `int64`          | Integer values                                  | `10, 20, 30`        |
| `float64`        | Decimal values                                  | `10.5, 20.75`       |
| `object`         | Often used for strings or mixed Python objects  | `"Hari", "Ravi"`    |
| `bool`           | Boolean values                                  | `True, False`       |
| `string`         | Text values using Pandas' string dtype          | `"Hello", "Python"` |
| `datetime64[ns]` | Date and time values                            | `2026-10-22`        |
| `category`       | Values belonging to a limited set of categories | `"Male", "Female"`  |


#### Checking Data Types Using dtypes

We can use the dtypes attribute to check the data type of each column in a DataFrame.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85.5, 90.0, 78.5],
    "Passed": [True, True, True]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

#### Checking the Data Type of a Single Column

Use the dtype attribute to check one column's data type.

```python
print(df["Age"].dtype)
print(df["Marks"].dtype)
print(df["Name"].dtype)
```

### 7. Index

An index in Pandas is a label used to identify and access rows in a Series or DataFrame. By default, Pandas assigns numerical index labels starting from 0, but we can also create custom indexes.

#### Default Index

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21]
}

df = pd.DataFrame(data)

print(df)
```

### Creating a Custom Index
We can assign meaningful labels using the index parameter.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21]
}

df = pd.DataFrame(
    data,
    index=["S101", "S102", "S103"]
)

print(df)
```

### 8. Columns

A column in Pandas represents a particular attribute or variable in a DataFrame. Each column has a name and contains values for that attribute.

```python
import pandas as pd

data = {
    "Name": ["Hari", "Ravi", "Kiran"],
    "Age": [22, 23, 21],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

**Accessing a Single Column**

```python
print(df["Name"])
```

**Accessing Multiple Columns**

```python
print(df[["Name", "Age"]])
```

**Updating an Existing Column**

```python
#You can modify the values in an existing column by assigning new values.

df["Age"] = df["Age"] + 1
print(df[["Name", "Age"]])
```

**Renaming Columns**

```python
#Use rename() to change column names.

df.rename(
    columns={"Name": "Student_Name", "Marks": "Total_Marks"},
    inplace=True
)

print(df.columns) 
```

**Deleting a Column**

```python
df.drop(columns=["City"], inplace=True)

print(df.columns)
```

### 9. Reading Data

Reading data in Pandas is the process of loading data from external files or data sources into a Pandas DataFrame so that we can analyze, clean, and manipulate it using Python.

Pandas supports reading data from different file formats, including CSV, Excel, JSON, and SQL databases.



#### Reading a CSV File — pd.read_csv()

A CSV (Comma-Separated Values) file stores data in a tabular format, with values separated by commas.

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)

##Reading Only a Few Rows

import pandas as pd

df = pd.read_csv("students.csv", nrows=2)

print(df)

##Reading Specific Columns

import pandas as pd

df = pd.read_csv("students.csv", usecols=["Name", "Marks"])

print(df)

##Reading a CSV File from a Specific Path

import pandas as pd

df = pd.read_csv(r"C:\Users\YourName\Documents\students.csv")

print(df)
```

#### Reading an Excel File — pd.read_excel()

Pandas can read Excel files with .xlsx or .xls extensions.

```python
import pandas as pd

df = pd.read_excel("students.xlsx")

print(df)
```

#### Reading a JSON File — pd.read_json()

JSON (JavaScript Object Notation) is a format commonly used to store and exchange structured data.

```python
import pandas as pd

df = pd.read_json("students.json")

print(df)
```
